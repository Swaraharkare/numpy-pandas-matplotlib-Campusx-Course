###`Q-9`
    i. Read `items.csv` making `item_name` as index.
    ii. Show no of nan values
    ii. Item price is given in $, so convert it to rupees without currency symbol.
    iii. Make data type of newly made series as float.
    iv. Fill nan with mean of the series


How csv file looks

```
item_name	item_price
Chips and Fresh Tomato Salsa	$2.39
Izze	$3.39
Nantucket Nectar	$3.39
Chips and Tomatillo-Green Chili Salsa	$2.39
Chicken Bowl	$16.98

```

````python
# code here
series1=series['item_price']
print('no of nan values',series1.isnull().sum())
````

````python
#. Item price is given in $, so convert it to rupees without currency symbol. 
clean_strings = series1.str.replace('$', '', regex=False).str.strip()

# 2. Convert to numeric, turning unparseable text into NaN
usd_numeric = pd.to_numeric(clean_strings, errors='coerce')

# 3. Multiply by the exchange rate to get Rupees
exchange_rate = 84.50
rupees_series = usd_numeric * exchange_rate

print(rupees_series)
````

````python
#Make data type of newly made series as float.
print(rupees_series.astype(float))
````

````python
#Fill nan with mean of the series
rupees_series.fillna(rupees_series.mean())
````

###`Q-10`:
    i. Find mean price
    ii. Find 30th and 6th percentile value
    iii. Plot Histogram on price with bin size 50
    iv. No of items price lies between [1000 to 2000]



````python
# code here
print(rupees_series.mean())
````

````python
percentile_30 = rupees_series.quantile(0.30)
percentile_6 = rupees_series.quantile(0.06)

print(f"30th Percentile: {percentile_30}")
print(f"6th Percentile:  {percentile_6}")
````

````python
#Plot Histogram on price with bin size 50
clean_prices = rupees_series.dropna()

min_val = int(np.floor(clean_prices.min() / 50) * 50)
max_val = int(np.ceil(clean_prices.max() / 50) * 50)
bin_edges = range(min_val, max_val + 50, 50)

# 3. Plot using the calculated bin ranges
clean_prices.hist(bins=bin_edges, edgecolor='black', grid=False)

# 4. Add labels and display layout
plt.title("Distribution of Item Prices (Bin Size: 50)")
plt.xlabel("Price in Rupees (INR)")
plt.ylabel("Frequency")
plt.show()
````
