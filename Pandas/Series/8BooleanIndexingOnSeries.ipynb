```python
df = pd.read_csv("movies.csv",index_col='Movie_Title')
# Convert to Series
movies = df['Lead_Actor']
print(type(movies))
print(movies)
```

```python
#counts movies of each actor
num_movies=movies.value_counts()
num_movies
```

```python
num_movies_cnt=num_movies[num_movies>2]
num_movies_cnt
```

```python
plt.bar(num_movies_cnt.index,num_movies_cnt,color='blue', edgecolor='black')
plt.title('Number of Movies', fontsize=14, fontweight='bold')
plt.xlabel('Category')
plt.ylabel('Count')

plt.xticks(rotation=45, ha='right')
plt.show()
```
