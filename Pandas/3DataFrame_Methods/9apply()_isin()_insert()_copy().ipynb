```python
points_df = pd.DataFrame(
    {
        '1st point':[(3,4),(-6,5),(0,0),(-10,1),(4,5)],
        '2nd point':[(-3,4),(0,0),(2,2),(10,10),(1,1)]
    }
)

points_df
```

```python
def euclidean(row):
  pt_A = row['1st point']
  pt_B = row['2nd point']

  return ((pt_A[0] - pt_B[0])**2 + (pt_A[1] - pt_B[1])**2)**0.5
```

```python
points_df['distance'] = points_df.apply(euclidean,axis=1)
points_df
```

```python
# isin(series)
import pandas as pd

# Your main DataFrame
df = pd.DataFrame({
    'player': ['V Kohli', 'MS Dhoni', 'RG Sharma', 'G Maxwell'],
    'team': ['RCB', 'CSK', 'MI', 'RCB']
})

# The Series containing values you want to match
selected_players = pd.Series(['V Kohli', 'RG Sharma'])

# Filter the DataFrame
filtered_df = df[df['player'].isin(selected_players)]
print(filtered_df)
```

```python
# copy(series + dataframe)

#The df.copy() method in pandas is used to create a completely independent clone of a DataFrame or a Series.
#df.copy(deep=True)

df1 = pd.DataFrame({'A': [1, 2, 3]})
df2 = df1.copy()  # Creates a totally independent clone

df2['A'] = [99, 99, 99]

print(df1['A'].tolist())  
# Output: [1, 2, 3] <-- Safe! df1 remains untouched.
print(df2['A'].tolist())  
```
