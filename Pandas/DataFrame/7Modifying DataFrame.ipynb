```python
import numpy as np
import pandas as pd
```

```python
#Renaming Column Labels in a DataFrame

# Create a DataFrame
df = pd.DataFrame({'A': [1, 2, 3],'B': [4, 5, 6]})
print(df)

# Rename column 'A' to 'aa'
df = df.rename(columns={'A': 'aa'})

# Display modified DataFrame
print(df)
```

```python
#Renaming Row Labels in a DataFrame

# Create a DataFrame
df = pd.DataFrame({'A': [1, 2, 3],'B': [4, 5, 6]}, index=['x', 'y', 'z'])

# Display original DataFrame
print(df)

# Rename the multiple row labels
df = df.rename(index={'x': 'r1', 'y':'r2', 'z':'r3'})

# Display modified DataFrame
print(df)
```

```python
#Adding or Inserting Columns in a DataFrame

# Create a DataFrame
df = pd.DataFrame({'A': [1, 2, 3],'B': [4, 5, 6]})

# Add a new column 'C' with values
df['C'] = [7, 8, 9]

# Display updated DataFrame
print(df)
```

```python
#Inserting a Column at a Specific Position
# Create a DataFrame
df = pd.DataFrame({'A': [1, 2, 3],'B': [4, 5, 6]})

# Insert a new column 'D' at position 1
df.insert(1, 'D', [10, 11, 12])

# Display updated DataFrame
print(df)
```

```python
#Replacing the Contents of a DataFrame

# Create a DataFrame
df = pd.DataFrame({'A': [1, 2, 3],'B': [4, 5, 6]})

# Replace the contents of column 'A' with new values
df['A'] = [10, 20, 30]

# Display updated DataFrame
print(df)
```

```python
#Replacing Specific Values using the replace() method
# Create a DataFrame
df = pd.DataFrame({'A': [1, 2, 3],'B': [4, 5, 6]})

# Display the Input DataFrame 
print("Original DataFrame:", df, sep='\n')

# Replace the contents 
df.replace({'A': 1, 'B': 6}, 100, inplace=True)

# Display updated DataFrame
print(df)
```

```python
#Deleting Columns

# Create a DataFrame
df = pd.DataFrame({'A': [1, 2, 3],'B': [4, 5, 6],'C': [7, 8, 9]})

# Display the original DataFrame
print("Original DataFrame:", df, sep='\n')

# Delete columns 'A' and 'B'
df = df.drop(columns=['A', 'B'])

# Display updated DataFrame
print(df)
```

# Removing Rows from a DataFrame

- Removing the irrelevant data

- Removing duplicate or missing values

- Deleting specific rows based on conditions

1) Using the .drop() method

2) Removing rows based on conditions

3) Dropping rows with index slicing

```python
#Dropping Rows using the drop() method--------DataFrame.drop() method based on its index value.

df = pd.DataFrame({'A': [1, 2, 3, 4, 5],'B': [4, 5, 6, 7, 8]})

# Display original DataFrame
print(df)

# Drop the row with index 3
result = df.drop(3)

# Display the result
print(result)
```

```python
#Dropping Multiple Rows by Labels-----

df = pd.DataFrame({'A': [1, 2, 3, 4, 5],'B': [4, 5, 6, 7, 8],
'C': [9, 10, 11, 12, 13]}, index=['r1', 'r2', 'r3', 'r4', 'r5'])

# Display original DataFrame
print(df)

# Drop the rows by row-labels
result = df.drop(['r1', 'r3'])

# Display the result
print(result)
```

```python
#Removing Rows Based on a Condition

# Create a DataFrame
df = pd.DataFrame({'A': [1, 2, 3, 4, 5],'B': [4, 5, 6, 7, 8],
'C': [90, 0, 11, 12, 13]}, index=['r1', 'r2', 'r3', 'r4', 'r5'])

# Display original DataFrame
print(df)

# Dropping rows where column 'C' contains 0
result = df[df["C"] != 0]   
#When you pass this Boolean mask back into the DataFrame using square brackets df[...], 
#pandas keeps only the rows where the condition is True and drops the rows where it is False.

# Display the result
print(result)
```

```python
#Removing Rows using Index Slicing

# Create a DataFrame
df = pd.DataFrame({'A': [1, 2, 3, 4, 5],'B': [4, 5, 6, 7, 8]})

# Display original DataFrame
print(df)

# Drop the row using the index slicing
result = df.drop(df.index[2:4])

# Display the result
print(result)
```
