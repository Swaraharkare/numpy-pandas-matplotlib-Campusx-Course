```python
b=np.arange(10).reshape(5,2)
b
```

```python
#np.append() adds values to the end of an existing array.
```

```python
#to append 1 in 2d array --add in column at last
print(np.append(b,np.ones((b.shape[0],1)),axis=1))
```

```python
#Finds the number of rows in your original array b
b.shape[0]

#Creates a new, single-column 2D array of ones with the exact same number of rows as b
np.ones((b.shape[0], 1))

#axis=1: Specifies that the two arrays should be joined horizontally (column-wise).
```

```python
#to append random in 2d array --add in column at last
print(np.append(b,np.random.random((b.shape[0],1)),axis=1))
```

```python
#add another array elements at last column
c=np.arange(10,15).reshape(5,1)
print(np.append(b,c,axis=1))
```
