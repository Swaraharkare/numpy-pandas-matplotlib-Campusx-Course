```python
import numpy as np

a=np.array([1,1,2,2,3,4,5,5,5,6,7])
np.unique(a)
```

```python
arr = np.array([5, 2, 1, 2, 5, 3, 1, 4])
unique_elements = np.unique(arr)

print(unique_elements)
```

```python
arr = np.array(['apple', 'banana', 'apple', 'orange', 'banana', 'apple'])
elements, counts = np.unique(arr, return_counts=True)

print("Elements:", elements)  # Output: ['apple' 'banana' 'orange']
print("Counts:  ", counts)    # Output: [3 2 1]
```
