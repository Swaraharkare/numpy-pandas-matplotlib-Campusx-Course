```python
import numpy as np
```

```python
a= np.array([11, 53, 28, 50, 38, 37, 94, 92,  5, 30, 68,  9, 78,  2, 21])
a
```

```python
# find all indices with value greater than 50
np.where(a>50)
```

```python
# replace all values > 50 with 0
np.where(a>50,0,a)
```

```python
np.where(a%2 == 0,0,a)
```

```python
matrix = np.array([[8,4],
    [9, 2]
])

# Find rows and columns where elements equal 9 or 8
rows, cols = np.where(matrix > 5)

print("Rows:", rows)  # Output: [0, 1]
print("Cols:", cols)  # Output: [0, 0]
# The values match positions (0,1) and (0,0)
```
